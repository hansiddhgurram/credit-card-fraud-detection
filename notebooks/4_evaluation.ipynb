```python
import joblib
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score
from sklearn.metrics import precision_recall_curve
```

```python
from sklearn.ensemble import RandomForestClassifier

X_train, y_train = joblib.load("../data/train_processed.pkl")
X_test, y_test = joblib.load("../data/test_processed.pkl")

rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

y_probs = rf_model.predict_proba(X_test)[:, 1]
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_probs)
auc_score = roc_auc_score(y_test, y_probs)

plt.figure()
plt.plot(fpr, tpr)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title(f"ROC Curve (AUC = {auc_score:.4f})")
plt.show()

print("ROC-AUC Score:", auc_score)
```

```python
precision, recall, thresholds_pr = precision_recall_curve(y_test, y_probs)

plt.figure()
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.show()
```

```python
def predict_with_threshold(probs, threshold):
    return (probs >= threshold).astype(int)
```

```python
from sklearn.metrics import classification_report

for t in [0.3, 0.4, 0.5, 0.6, 0.7]:
    print(f"\nThreshold: {t}")
    y_pred_custom = predict_with_threshold(y_probs, t)
    print(classification_report(y_test, y_pred_custom))
```
